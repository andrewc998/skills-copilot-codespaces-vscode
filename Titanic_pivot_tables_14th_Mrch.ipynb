```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_titanic = pd.read_csv('/content/drive/MyDrive/titanic.csv')
df_titanic.head()
```

```python
df_titanic.columns
```

```python
#remove duplicates
df_titanic = df_titanic.drop_duplicates()
```

```python
#remove null values
df_titanic = df_titanic.dropna()
```

```python
df_titanic.info()
```

```python
#group by 'sex' and find average fare for each sex
df_titanic.groupby('Sex')['Fare'].describe()
```

```python
# shows that male were many
df_titanic['Sex'].value_counts()
```

```python
#groupby sex and pclass to get aggregate
#shows that despite females having highest mean fare, male were many
# might be issue of unclean data, zeros affecting values overall

df_titanic.groupby(['Sex','Pclass'])['Fare'].describe()
```

```python
#https://github.com/learn-co-curriculum/dsc-pivot-tables-pandas
```

```python
#PIVOT TABLES
#uses dataframe.pivot()
#takes 3 parameters:
# 1 index - column that u want to use as new index #1
# 2 column - columns to be used as new columns - this is what you convert to wide format
# values -  what data we expect to be looking at

#EG some_dataframe.pivot(index='State', columns='Gender', values='Deaths_mean')

# it is key to pivot with a categorical column rather than a continous/ numerical column
```

```python
df_titanic.head(3)
```

```python
df_titanic.pivot(columns='Sex', values=['Age', 'Fare']).describe()
```

```python
#pivot table to count number of passangers per class
df_titanic.pivot_table(columns='Pclass', index='Survived', values='Name', aggfunc='count')
```

```python
# pivot table for average age of each survival group
#pivot table of survived
df_titanic.pivot_table(columns ='Survived', values='Age', aggfunc='mean')
```

```python
# PIVOT TABLES WITH AGGREGATE FUNCTIONS
# - involves just asking for multiple functions eg sum,count etc
# use .agg(mean......)
```

```python
df_titanic.pivot(columns ='Survived', values='Age',).agg(['mean','min','max','count'])
```

```python
# PIVOT MULTIPLE COLUMNS
```

```python
df_titanic.pivot( columns=['Sex','Pclass'], values=['Age','Fare']).describe()
```
