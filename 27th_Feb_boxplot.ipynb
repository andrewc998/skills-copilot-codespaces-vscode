HISTOGRAM

Visual that shows distribution of values in a dataset.
Made of bars and frequencies where:

X AXIS rep BINS

Y AXIS rep FREQUENCIES

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#load an in built dataset
data_plot = sns.load_dataset('tips')
print(data_plot)
data_plot.head(20)
```

```python
plt.figure(figsize = (10 ,8)) #uses matplotlib  to plot size
sns.histplot(data_plot['total_bill'], #uses histogram and its functions to display data
             bins = 20, color = 'green')
plt.title("Total bill Histogram Distribution")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
#BoxPlot is visual data that summarises data in 5 categories
#Minimum:Lowest value(excludes outliers)
#first quartile = 25th percentile (Q1)
#Median = 50th percentile (Q2)
#third quartile = 75th percentile (Q3)
#Maximum: Highest value(excludes outliers)

#It also includes whiskers = lines extendimg from the box and alsp possible outliers(plotted as dots outside the whiskers)
```

```python
color_map = {'Thur': 'green', 'Fri': 'blue', 'Sat': 'grey', 'Sun': 'yellow'}
plt.figure(figsize = (10,8))
sns.boxplot(x = 'day', y = 'tip', palette = color_map, data = data_plot) # we have not told it like the line graph what data to use
plt.title ('Tips per day')
plt.xlabel('Day')
plt.ylabel('Tip')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.boxplot(x='day', y='tip', hue = 'day', palette= 'coolwarm', data=data_plot)  # Use a palette for different colors
plt.title('Tips per day')   # notice we have not told it like the line graph what data to use from above
plt.xlabel('Day')
plt.ylabel('Tip')
plt.show()
```

Scatter plot

Visual that shows r/ship btn 2 variables. Each point represents an observation

X shows one var while Y shows another variable

```python
plt.figure(figsize=(10, 8))
sns.regplot(x = 'total_bill', y = 'tip', data = data_plot, line_kws={'color': 'green'},
            scatter_kws={'alpha':0.5})
plt.title('Scatter Plot Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```
