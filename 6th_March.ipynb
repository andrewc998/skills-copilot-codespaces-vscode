```python
def load_csv(path="/content/apple_orchard.csv"):
  import csv
  with open(path, newline='') as f:
    reader = csv.DictReader(f)
    # Use a different variable name to avoid conflict with built-in 'list' function
    data_list = [] # Initialize an empty list using []
    for row in reader: # Iterate over rows in reader
      data_list.append(row)  #add each row to the list
    return data_list

# Load the data from the CSV file
data = load_csv(path="/content/apple_orchard.csv")

# Print the data
# Option 1: Print the entire list of dictionaries
for row in data:
  print(row)
```

```python
new_data = load_csv(path="/content/apple_orchard.csv")
total_sum = 0
for row in new_data:
  total_sum += float(row['yield'])
avg = total_sum / len(new_data)
minimum = min(new_data, key=lambda x: x['yield'])
maximum = max(new_data, key=lambda x: x['yield'])
print(f"Average yield: {avg}")
print(f"Minimum yield: {minimum}")
print(f"Maximum yield: {maximum}")
```
